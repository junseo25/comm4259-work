## Which state's customers spend the most? (12 points)
The code is given. Run it, then answer the three questions in the cell after it.

````python
# Total, average and number of customers' spending (TotalSales_sum) for each state
state_sales = customer.groupby("State").TotalSales_sum.agg(["sum", "mean", "count"])
state_sales.sort_values("mean", ascending=False).head()
````

Answer here:
1. In your own words, what do `agg(["sum", "mean", "count"])` and `sort_values("sum", ascending=False)` do?
`agg(["sum", "mean", "count"])` summarizes and aggregates the three mathematical functions of sum, mean, and count related to the prevailing group, in this case, the state meaning that all numbers that are within that group are aggregated by those three mathematical functions. For example, all sales in CA are added together to find the sum, same with those labeled MI, or FL specifically by the group they are in. `sort_values("sum", ascending=False)` sorts the table from the highest to lowest when looking at the values in sum.

2. Which state's customers spend the most in total?
CA's customers spend the most in total.

3. Change `"sum"` to `"mean"` in the last line and run the cell again. Does the answer change, and which measure would you report to a manager?
The answer does not change, CA's customers still spend the most in total, but on average a customer from AK spends the most. Depending on the report that is being asked from the manager, this would change. For example, if they would like to know where to target marketing spend to maximize value, it is possible that the mean would be more useful while wanting to know where the company has been most successful might have sum being more useful. It is important to note the counts here as counts that are way below or above average may indicate this being an outlier and needing to be investigated further.

**Example** (Customer data, not Titanic): the feature **AgeAtJoin** = year of JoinDate minus BirthYear. Hypothesis: older customers churn less, because they may be more settled in their choices. Check with `customer.groupby("Churn").AgeAtJoin.mean()`: the average AgeAtJoin is 60.9 for returning customers and 60.4 for churned customers.
````python
# Feature 1, worked example: FamilySize = the passenger + siblings/spouses (SibSp) + parents/children (Parch)
titanic["FamilySize"] = titanic["SibSp"] + titanic["Parch"] + 1

# Survival rate for each family size, and how many passengers have each size
print(titanic.groupby("FamilySize").Survived.mean().round(3))
print(titanic["FamilySize"].value_counts().sort_index())
````

Your explanation of the worked example (what each line does, the hypothesis, and what the data shows):

Explanation of code: The first line creates a new column in the dataset by adding the number of siblings and spouses, the parents and children, and the passenger themselves. It then groups the passengers by their family size and calculates the average survival rate for the passenger dependent on their family's size rounded to three decimal places. As the resulting number is binary with having survived being 1 and not being 0, the mean represents the proportion of the family size that survived. It then provides a count of the passenger based on the family size of the passenger.

Hypothesis: Passengers with small families are more likely to survive than those with no or large families as small families could help one another quickly while those without families would find it difficult to find similar support and larger families could find it difficult to stay together.

Result: There is strong evidence that there is a relationship between family size and survival with those who traveled alone averaging a 30.4% survival rate, those with a family size between 2 and 4 averaging a survival rate of 55.3%, 57.8%, and 72.4% respectively while those with families of 5 or larger had lower survival rates than those with smaller family sizes.

**Sample patterns for Feature 2** (Customer Churn data, not Titanic). Most new features follow one of the four patterns below. Run the cell to see each one, then adapt the pattern that fits your idea in the Feature 2 cell. Pattern 4, taking part of a text value, is the one most students find hardest: split the text, keep one piece, and split again.

````python
# SAMPLE (Customer Churn data, not Titanic). Each pattern creates a column, then checks it with a groupby.
customer = pd.read_csv("CustomerChurn_cleaned.csv")
customer["Churned"] = (customer["Churn"] == "NotRetained").astype(int)   # 1 = churned, so .mean() is a churn rate

# Pattern 1: a 0/1 flag from a text column
customer["EmailOK"] = (customer["FlagEmailable"] == "Y").astype(int)
print(customer.groupby("EmailOK").Churned.agg(["mean", "count"]).round(3))

# Pattern 2: a ratio of two number columns
customer["SalesPerOrder"] = customer["TotalSales_sum"] / customer["Orders_count"]
print(customer.groupby("Churned").SalesPerOrder.mean().round(1))

# Pattern 3: groups (bins) of a number column, with pd.cut
customer["SpendLevel"] = pd.cut(customer["TotalSales_sum"], bins=[0, 100, 500, 100000], labels=["low", "mid", "high"])
print(customer.groupby("SpendLevel", observed=True).Churned.agg(["mean", "count"]).round(3))

# Pattern 4: part of a text value. "2012-01-03 16:36:00": split at the space and keep piece 0 ("2012-01-03"),
# then split at "-" and keep piece 0 ("2012")
customer["JoinYear"] = customer["CustomerSinceDate"].str.split(" ").str[0].str.split("-").str[0]
print(customer.groupby("JoinYear").Churned.agg(["mean", "count"]).round(3))
````

````python
# Feature 2 (yours): create it and check it with a groupby, then save both features with to_csv()

#Feature 2
titanic["IsChild"] = (titanic["Age"] < 18).astype(int)

# Survival rate for children, and how many children there are on the Titanic
print(titanic.groupby("IsChild").Survived.mean().round(3))
print(titanic["IsChild"].value_counts().sort_index())

titanic.to_csv("titanic_features.csv", index=False)
````

I then used the groupby function to group passengers with this variable and calculated the average survival rate for children and non-children using .Survived.mean(). Alongside this, I used the value_counts() function to see how many passengers were in each group.
